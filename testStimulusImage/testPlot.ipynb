# Test ImageStimulus updates

```
import warnings

import matplotlib.pyplot as plt
import numpy as np
import polars as pl

import pymovements as pm
from pymovements.gaze.experiment import Experiment
from pymovements.stimulus import ImageStimulus
```

## Set up

```
experiment = Experiment(
    screen_width_px=1280,
    screen_height_px=1024,
    screen_width_cm=38,
    screen_height_cm=30.2,
    distance_cm=68,
    origin="upper left",
    sampling_rate=250.0,
)

# Load gaze data from a CSV file and initialize three Gaze objects to test different modes.
gaze1 = pm.gaze.from_csv(
    "./gaze-toy-example.csv",
    experiment=experiment,
    time_column="time",
    pixel_columns=["x", "y"],
)

stimIm = "..\\tests\\files\\stimuli\\pexels-zoorg-1000498.jpg"
```

# Plot the stimulus image  

```
stim = ImageStimulus.from_file(stimIm)

fig, ax = stim.plot(0)

ax.set_title("Stimulus image")

plt.show()
```

# Plot the stimulus image on already defined axes

```
stim = ImageStimulus.from_file(stimIm)

fig, ax = plt.subplots()

stim.plot(0, ax=ax)

ax.set_title("Stimulus on existing axes")
plt.show()
```

# Plot the gaze over the stimulus image

```
fig, ax = plt.subplots()

stim = ImageStimulus.from_file(stimIm)

fig, ax = stim.plot(0, ax=ax)

pm.plotting.traceplot(gaze1, ax=ax)

ax.set_title("Traceplot on stimulus image")

plt.show()
```

# Check deprecation of old stimulus image parameters

```
# with warnings.catch_warnings(record=True) as w:
#    warnings.simplefilter("always")
warnings.simplefilter("always", DeprecationWarning)
fig, ax = pm.plotting.traceplot(
    gaze1,
    add_stimulus=True,
    path_to_image_stimulus=stimIm,
)

# print([str(wi.message) for wi in w])
```
